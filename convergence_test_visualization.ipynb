# CONVERGENCE TEST VISUALIZER

This code is used to visualize the results produced by the converge tests made in the GR-EM-KG simulator. 

```python
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact, fixed, IntSlider



#del fitxer functions_aux
from functions_aux import lector_fitxers
from functions_aux import info_converg
from functions_aux import pintar_fotograma
```

```python
#MODIFICAR DOMÉS AIXO
#---------------------------------------
#amb això triam sa carpeta i sa simulació que volem comprovar convergència
simulacio = "potential4_lam100_mu200_m1"
#---------------------------------------

scaler = 3 #o bé 1.5 o bé 3



#així hem definit sa ruta genèrica de sa nostra simulació
ruta = "conv_test/" + simulacio
```

### Visualització de H (cas scaler 1.5)

```python
variable    = "constr_H_c4"
constraint   = 1

x, t, results_1, results_2, results_3, results_4 = info_converg(variable, ruta, constraint=constraint, scaler=scaler)

interact(
    pintar_fotograma, 
    i=IntSlider(min=0, max=len(t)-1, step=1),  # El slider ahora mueve el tiempo 'i'
    variable=fixed(variable),                  # Fijamos el texto
    constraint=fixed(constraint), 
    x=fixed(x),
    t=fixed(t),
    results_1=fixed(results_1),
    results_2=fixed(results_2),
    results_3=fixed(results_3),      
    results_4=fixed(results_4),
    scaler=fixed(scaler)                                # Esto salva a interact del ValueError
)
```
